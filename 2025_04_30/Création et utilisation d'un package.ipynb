# Package

Un package est un répértoire qui contient de modules ou des sous packages. 

Le fichier ``__init__.py`` sera importé par le import nom_du_package

```python
import package
```

```python
from package.mod1 import a
from package import mod1
```

```python
import package.mod3
```

```python
package.mod3.c
```

On peut restructurer un module en un package pour séparer les fonctionnalités

(J'ai utilisé un autre nom pour qu'il n'y ait pas de confusion)

```python
from bipolar_circuit import Resistor, Inductor, Capacitor

R1 = Resistor(10)
R2 = Resistor(5)
L1 = Inductor(15E-6)
C1 = Capacitor(10E-6)
circuit = (C1|L1|R1) + R2

print(circuit)
circuit.impedance(50)
```

Le fichier setup.py permet de regrouper les informations d'un package pour pouvoir le distribuer et l'installer: 

* python setup.py sdist
* python setup.py install

Il est aussi possible d'utiliser pip : 

* pip install .

(le . signifie que l'on prend le setup.py dans le repertoire courant)
